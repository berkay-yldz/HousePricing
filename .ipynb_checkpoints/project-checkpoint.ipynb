# "... TL" Temizleme İşlemi

```python
def clean_price(val):
    try:
        if pd.isna(val):
            return np.nan
        val_str = str(val)
        # İçindeki ilk sayıyı, ondalıklı da olabilir, yakala
        match = re.search(r"\d+[\.,]?\d*", val_str.replace('.', '').replace(',', ''))
        if match:
            return float(match.group(0))
        else:
            return np.nan
    except:
        return np.nan
```

```python
df.price = df.price.apply(clean_price)
```

```python
df.price
```

# "... m2" Temizleme İşlemi

```python
def clean_m2(val):
    try:
        if pd.isna(val):
            return np.nan
        val_str = str(val)
        match = re.search(r"\d+[\.,]?\d*", val_str)
        if match:
            return float(match.group(0).replace(",", "."))
        else:
            return np.nan
    except:
        return np.nan
```

```python
m2_cols = ['GrossSquareMeters','NetSquareMeters','HallSquareMeters','WCSquareMeters','BathroomSquareMeters','BalconySquareMeters']
```

```python
for col in m2_cols:
    clean_col_name=f"{col}_clean"
    df[clean_col_name]=df[col].apply(clean_m2)
```

```python
df[[f"{col}_clean" for col in m2_cols]].head()
```

# Birim metrekare fiyatı (price_per_m2):

```python
df['price_per_m2'] = df.price / df.GrossSquareMeters_clean
```

# NumberOfRooms temizleme işlemi:

```python
df.NumberOfRooms.unique()
```

```python
def clean_number_of_rooms(val):
    try:
        if pd.isna(val):
            return np.nan
        val_str = str(val).strip()
        
        # Özel durum: Stüdyo → 1 kabul edelim
        if val_str.lower() == 'stüdyo':
            return 1
        
        # Özel durum: 8+ gibi yazımlar → 9 oda olarak alalım
        if re.match(r"^\d+\+\s*Oda$", val_str):
            num_part = int(val_str.split('+')[0])
            return num_part + 1  # 8+ → 9 gibi
        
        # X+Y formatını yakala (ör: 3+1, 2.5+1, 7+3, 4.5+1)
        match = re.match(r"(\d+\.?\d*)\s*\+\s*(\d+\.?\d*)", val_str)
        if match:
            room1 = float(match.group(1))
            room2 = float(match.group(2))
            return room1 + room2
        
        # X Oda formatını yakala (ör: 5 Oda, 1 Oda)
        match2 = re.match(r"(\d+\.?\d*)\s*Oda", val_str)
        if match2:
            return float(match2.group(1))
        
        # Eğer hiçbiri eşleşmezse → NaN döndür
        return np.nan
    
    except:
        return np.nan
```

```python
df["NumberOfRooms_clean"] = df["NumberOfRooms"].apply(clean_number_of_rooms)
```

```python
df.NumberOfRooms_clean # 8+ Oda değeri 9 kabul edildi 
```

# *~ Korelasyon*

## Korelasyon Nedir?
İki veya daha çok değişken arasındaki ilişkilere korelasyon denmektedir.
### Korelasyon Analizi
İki değişken arasında doğrusal ilişki olup olmadığını, varsa bunun yön ve gücünü ortaya çıkarmak için kullanılan bir istatistik tekniğidir.
    
*Örneğin:* 
- *Bir sınıftaki öğrencilerin final sınavı için ayırdıkları çalışma süreleri ile final sınav puanları arasında doğrusal bir ilişki bulunup bulunmadığını, ilişki varsa bunun yönü ve derecesini ortaya çıkarmak için yapılan bir çalışma korelasyon çalışmasıdır. Bu durumda çalışma süreleri ve sınav puanları birer değişkendir ve bu iki değişken arasındaki ilişki korelasyondur.*
     
### Amaç
 - Değişkenler arasında doğrusal bir ilişki var mı?
 - İlişki varsa bunun yönü ve gücü nedir?
 - Değişkenlerin birindeki değişm diğerinde ne kadar bir değişmeye yol açar?

*DİKKAT: Korelasyon analizi sonucuna göre her zaman neden-sonuç çıkarımında bulunmak, değişkenlerden birini, diğerinin nedeni olarak kabul etmek YANLIŞTIR.*

### Korelasyon Yönü ve Tipi
Korelasyon analizinde iki temel boyut dikkate alınır:
1) Yönü (Pozitif, Negatif, Nötr)
2) Tipi (Doğrusal, Doğrusal Olmayan)


Özet Tablo:

| Yön             | Tipi                 | Açıklama / Örnek                               |
| --------------- | -------------------- | ---------------------------------------------- |
| **Pozitif**     | **Doğrusal**         | Boy uzunluğu ile kilo artışı                   |
| **Negatif**     | **Doğrusal**         | Ders çalışma süresi ↑, hata sayısı ↓           |
| **Pozitif**     | **Doğrusal olmayan** | Motivasyon artışı ile başarı (doygunluk)       |
| **Negatif**     | **Doğrusal olmayan** | Stres düzeyi ↑, performans ↓ (karmaşık ilişki) |
| **Yok (Sıfır)** | **Herhangi**         | Rastgele ilişki yok                           |
